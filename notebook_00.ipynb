```r
library(glue)

usuario <- 'Megalonnix'
repositorio <- 'Projeto_Integrador_PI3_2Sem_2026'
nm_dataset_usado <- 'ipdm.csv'

# 1. Cria a URL correta com o "raw." no início e a barra antes do usuário
URL_GITHUB <- glue(
'https://raw.githubusercontent.com/{usuario}/{repositorio}/main/{nm_dataset_usado}')

# 2. Lê o arquivo CSV diretamente da internet
dados <- read.csv(URL_GITHUB, sep = ";", dec = ",", fileEncoding = 'Latin1')
```

```r
# 6. Controla melhor oqq eu quero ver:

getDadosFromMunicipio <- function(nmMunicipio, qtd_registros) {
 head(
  filter(dados, Municipio == nmMunicipio),
  n = qtd_registros)
}
```

```r
# 8. FILTRA OS DADOS, PEGA SÓ DA BAIXADA!!!!

cidades_baixada <- c("Santos", "São Vicente", "Guarujá", "Praia Grande",
                     "Cubatão", "Itanhaém", "Peruíbe", "Mongaguá", "Bertioga")

dados_baixada <- filter(dados,
                        Municipio %in% cidades_baixada)
```

```r
# lê 5 linhas de cada uma das cidades da BAIXADA (cada uma tem linha p/ cacete):

getDadosFrom_MunicipioBaixada <- function(nmMunicipio, qtd_registros) {
 head(
  filter(dados_baixada, Municipio == nmMunicipio),
  n = qtd_registros)
}

count <- 1
for (cidade in cidades_baixada) {
  cat("===========================================================================\n")
  cat(glue('Cidade Nº{count} | ( Nome = {cidade} )'))
  cat("\n===========================================================================\n\n")

  print(getDadosFrom_MunicipioBaixada(cidade, 3))
  cat('\n')
  count = count + 1
}
```
